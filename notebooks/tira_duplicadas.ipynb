```python
import pandas as pd
import glob
```

```python
files = sorted(glob.glob('../dados/selected/*.csv'))
files
```

```python
len(files)
```

```python
df = pd.read_csv(files[0], sep=';', encoding="latin1", on_bad_lines="skip", dtype="string")

for file in files[1:]:
    try:
        df_temp = pd.read_csv(file, sep=';', encoding="latin1", on_bad_lines="skip", dtype="string")
    except Exception as e:
        print(f"Erro ao ler o arquivo {file}: {e}")
        continue
    df = pd.concat([df, df_temp], ignore_index=True)

df
```

```python
df["SOLICITACAO"].nunique()
```

```python
df.columns
```

```python
import numpy as np

df['MENSAGEM_RESPOSTA'] = df['TIPO'].shift(-1)

next_row_is_broken = df['SOLICITACAO'].shift(-1).isna()
df.loc[~next_row_is_broken, 'MENSAGEM_RESPOSTA'] = np.nan

df_clean = df.dropna(subset=['SOLICITACAO']).copy()

df_clean = df_clean[df_clean['SOLICITACAO'] != '---------------------'].reset_index(drop=True)

df_clean
```

```python
df_clean[df_clean["SOLICITACAO"] == "7288459"]
```

```python
df_clean_no_dups = df_clean.drop_duplicates(subset=['SOLICITACAO', "DATA", "HORARIO"], keep='first').reset_index(drop=True)
df_clean_no_dups
```

```python
df_clean_no_dups[df_clean_no_dups["SOLICITACAO"].isna()]
```

```python
# Keep identifiers as strings: applying .str directly to mixed numeric/object
# columns turns numeric solicitation IDs into NaN.
df_clean_no_dups = df_clean_no_dups.apply(lambda col: col.astype("string").str.strip())

# Empty strings -> NULL
df_clean_no_dups = df_clean_no_dups.replace("", pd.NA)
```

```python
df_clean_no_dups["DATA_NASC"] = df_clean_no_dups["DATA_NASC"].astype("string").str.strip()
df_clean_no_dups.loc[df_clean_no_dups["DATA_NASC"].isin(["", "nan", "NaN"]), "DATA_NASC"] = None
```

```python
df_clean_no_dups[df_clean_no_dups["SOLICITACAO"].isna()]
```

```python
df_clean_no_dups
```

```python
df_clean_no_dups.to_csv('../dados/processed/cleaned_156_2017_2018.csv', sep=';',  na_rep="", index=False)
```
