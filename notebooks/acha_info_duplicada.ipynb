# Cartão de Crédito / Conta bancária / PIX

```python
import re
import pandas as pd

print("--- Starting Credit Card Extraction Process ---")
print("This script scans text columns in a dataset to find, validate, and mask credit card numbers.")

# 1. Regex to capture candidate numbers (13 to 19 digits, with optional spaces or hyphens)
cc_candidate_regex = r'\b(?:\d[ -]*?){13,19}\b'
print(f"\n[Step 1] Initialized Regex: Searching for 13 to 19 digit sequences (ignoring spaces/hyphens).")

# 2. Luhn Algorithm Validator
def passes_luhn(number_str: str) -> bool:
    """
    Validates a credit card number using the Luhn algorithm (mod 10).
    It doubles every second digit from the right and sums them up.
    If the total modulo 10 is 0, the card number is mathematically valid.
    """
    digits = [int(d) for d in re.sub(r'\D', '', number_str)]
    if not (13 <= len(digits) <= 19):
        return False
    checksum = 0
    reverse_digits = digits[::-1]
    for i, digit in enumerate(reverse_digits):
        if i % 2 == 1:
            doubled = digit * 2
            checksum += doubled - 9 if doubled > 9 else doubled
        else:
            checksum += digit
    return checksum % 10 == 0

# 3. Identify Brand (including Elo and Hipercard)
def identify_brand(pan: str) -> str:
    """
    Identifies the card brand based on its prefix (BIN/IIN).
    Includes Brazilian local networks like Elo and Hipercard, 
    as well as standard international networks (Visa, Mastercard, AmEx).
    """
    cleaned = re.sub(r'\D', '', pan)
    
    # Hipercard: 606282 or 3841
    if re.match(r'^(606282|3841\d{2})', cleaned):
        return 'Hipercard'
    
    # Elo BIN ranges
    elo_prefixes = (
        '401178', '401179', '431274', '438935', '451416', '457393', '457631', '457632',
        '504175', '627780', '636297', '636368'
    )
    if cleaned.startswith(elo_prefixes) or \
       any(start <= int(cleaned[:6]) <= end for start, end in [
           (506699, 506778), (509000, 509999), (650031, 650033), 
           (650035, 650051), (650405, 650439), (650485, 650538), 
           (650541, 650598), (650700, 650718), (650720, 650727), 
           (650901, 650978), (651652, 651674), (655000, 655019), 
           (655021, 655058)
       ]):
        return 'Elo'
        
    # Standard International Networks
    if re.match(r'^4', cleaned) and len(cleaned) in (13, 16, 19):
        return 'Visa'
    if (re.match(r'^(5[1-5])', cleaned) or (len(cleaned) >= 4 and 2221 <= int(cleaned[:4]) <= 2720)) and len(cleaned) == 16:
        return 'Mastercard'
    if re.match(r'^(34|37)', cleaned) and len(cleaned) == 15:
        return 'American Express'
        
    return 'Unknown/Generic Card'

# 4. Extract, Validate, and Mask
def extract_valid_cards(text: str) -> list[str]:
    """
    1. Extracts candidate strings using Regex.
    2. Validates them mathematically using the Luhn algorithm.
    3. Identifies the brand.
    4. Returns a masked version (e.g., Visa (****-****-****-1234)) to prevent data leaks.
    """
    if not isinstance(text, str):
        return []
    
    matches = re.findall(cc_candidate_regex, text)
    valid_cards = []
    
    for candidate in matches:
        raw_digits = re.sub(r'\D', '', candidate)
        if passes_luhn(raw_digits):
            brand = identify_brand(raw_digits)
            # Mask all but the last 4 digits for safe auditing/display
            masked = f"{brand} ({raw_digits})"
            valid_cards.append(masked)
            
    return valid_cards

print("[Step 2] Processing DataFrame columns. Functions loaded: passes_luhn(), identify_brand(), extract_valid_cards()")

# Assuming 'text_cols' and 'df' are already defined in your environment
# text_cols = ['DESCRICAO', 'OBSERVACAO', 'MENSAGEM_RESPOSTA', 'RESPOSTA_FINAL']

print("\n[Step 3] Scanning specific columns for matches and applying masks...")
for col in text_cols:
    print(f" -> Scanning text in column: '{col}'...")
    df[f'CARDS_FOUND_IN_{col}'] = df[col].apply(extract_valid_cards)

print("\n[Step 4] Filtering dataset to isolate only the rows where card numbers were detected...")
# Filter rows where at least one valid card was discovered in any of the checked columns
mask_cards = df[[f'CARDS_FOUND_IN_{col}' for col in text_cols]].apply(lambda row: any(len(c) > 0 for c in row), axis=1)
df_cards_found = df[mask_cards].copy()

# Display results
print(f"\n--- Process Complete ---")
print(f"Total rows flagged containing valid credit card numbers: {len(df_cards_found)}")
print("Displaying the flagged records with masked card data for safety:\n")

display_cols_cards = ['SOLICITACAO'] + [f'CARDS_FOUND_IN_{col}' for col in text_cols]
display(df_cards_found[display_cols_cards])
```

```python
import re
import pandas as pd

def extract_bank_accounts(text: str) -> list[str]:
    if not isinstance(text, str):
        return []
    
    valid_accounts = []
    
    # 1. Contextual Regex (High Confidence)
    # Looks for combinations of "Agência" (or "Ag") and "Conta" (or "CC", "C/C") separated by noise
    context_regex = r'(?i)\b(?:ag[eê]ncia|ag\.?)\s*:?\s*(\d{4}(?:-[0-9XxzZ])?)\b.{0,30}?(?:conta|c/?c|c/?p)\s*:?\s*(\d{4,12}(?:-[0-9XxzZ])?)\b'
    
    for ag, cc in re.findall(context_regex, text):
        masked_cc = f"****-{cc.split('-')[-1]}" if '-' in cc else f"****{cc[-3:]}"
        valid_accounts.append(f"Ag: {ag} | Conta: {masked_cc} (Contextual Match)")

    # 2. Standalone Format Regex (Institution-Specific)
    # Looks for exact formatting structures of major banks, even if keywords aren't present.
    format_regexes = {
        'Banco do Brasil': r'\b(\d{4}-[0-9Xx])\s*[-/|]?\s*(\d{8}-[0-9Xx])\b',
        'Itaú': r'\b(\d{4})\s*[-/|]?\s*(\d{5}-\d)\b',
        'Bradesco': r'\b(\d{4})\s*[-/|]?\s*(\d{7}-\d)\b',
        'Santander': r'\b(\d{4})\s*[-/|]?\s*(\d{8}-\d)\b',
        'Caixa': r'\b(\d{4})\s*[-/|]?\s*(\d{3})\s*[-/|]?\s*(\d{8,9}-\d)\b' # Includes 3-digit Op Code (e.g., 001, 013)
    }
    
    # Only run standalone regexes if no contextual matches were found to prevent duplicates
    if not valid_accounts:
        for bank, pattern in format_regexes.items():
            for match in re.findall(pattern, text):
                # Caixa has 3 capture groups (Ag, Op, CC), others have 2 (Ag, CC)
                if len(match) == 3:
                    ag, op, cc = match
                    masked_cc = f"****-{cc[-1]}"
                    valid_accounts.append(f"{bank} -> Ag: {ag} | Op: {op} | Conta: {cc}")
                else:
                    ag, cc = match
                    masked_cc = f"****-{cc[-1]}"
                    valid_accounts.append(f"{bank} -> Ag: {ag} | Conta: {cc}")
                    
    return valid_accounts

for col in text_cols:
    df[f'BANK_ACCOUNTS_IN_{col}'] = df[col].apply(extract_bank_accounts)

# Filter rows where at least one bank account was discovered
mask_banks = df[[f'BANK_ACCOUNTS_IN_{col}' for col in text_cols]].apply(lambda row: any(len(c) > 0 for c in row), axis=1)
df_banks_found = df[mask_banks].copy()

# Display results
print(f"Found {len(df_banks_found)} rows containing bank account numbers.")
display_cols_banks = ['SOLICITACAO'] + [f'BANK_ACCOUNTS_IN_{col}' for col in text_cols]
display(df_banks_found[display_cols_banks])
```

# CNH / Título de Eleitor / PIS/PASEP / NIS

```python
import re
import pandas as pd

# 1. Regex for 11 digits (allows optional spaces, dots, or hyphens)
cnh_candidate_regex = r'\b(?:\d[.\-\s]*){11}\b'

# 2. CNH Algorithm Validator (Double Modulo 11)
def is_valid_cnh(cnh_str: str) -> bool:
    # Clean non-digits
    cnh = re.sub(r'\D', '', cnh_str)
    
    # Must be exactly 11 digits and cannot be a sequence of the same number
    if len(cnh) != 11 or cnh == cnh[0] * 11:
        return False

    try:
        # Calculate First Check Digit (10th digit)
        # Weights: 9, 8, 7, 6, 5, 4, 3, 2, 1
        sum1 = sum(int(cnh[i]) * (9 - i) for i in range(9))
        dv1 = sum1 % 11
        
        # CNH has a unique rule where the first check can generate an increment for the second
        inc = -2 if dv1 >= 10 else 0
        if dv1 >= 10:
            dv1 = 0

        # Calculate Second Check Digit (11th digit)
        # Weights: 1, 2, 3, 4, 5, 6, 7, 8, 9
        sum2 = sum(int(cnh[i]) * (1 + i) for i in range(9))
        dv2 = (sum2 % 11) + inc
        
        if dv2 < 0:
            dv2 += 11
        if dv2 >= 10:
            dv2 = 0

        # Compare calculated digits with the last two digits of the input
        return cnh[-2:] == f"{dv1}{dv2}"
        
    except ValueError:
        return False

# 3. Extract, Validate, and Mask
def extract_valid_cnhs(text: str) -> list[str]:
    if not isinstance(text, str):
        return []
    
    matches = re.findall(cnh_candidate_regex, text)
    valid_cnhs = []
    
    for candidate in matches:
        raw_digits = re.sub(r'\D', '', candidate)
        if is_valid_cnh(raw_digits):
            # Mask the CNH, showing only the last 4 digits
            masked = f"CNH: *******{raw_digits[-4:]}"
            valid_cnhs.append(masked)
            
    return valid_cnhs

for col in text_cols:
    df[f'CNH_FOUND_IN_{col}'] = df[col].apply(extract_valid_cnhs)

# Filter rows where at least one valid CNH was discovered
mask_cnh = df[[f'CNH_FOUND_IN_{col}' for col in text_cols]].apply(lambda row: any(len(c) > 0 for c in row), axis=1)
df_cnh_found = df[mask_cnh].copy()

# Display results
print(f"Found {len(df_cnh_found)} rows containing CNH numbers.")
display_cols_cnh = ['SOLICITACAO'] + [f'CNH_FOUND_IN_{col}' for col in text_cols]
display(df_cnh_found[display_cols_cnh])
```

```python
import re
import pandas as pd

# 1. Regex for 12 digits (allows optional spaces, dots, or hyphens)
titulo_candidate_regex = r'\b(?:\d[.\-\s]*){12}\b'

# 2. Título de Eleitor Validator
def is_valid_titulo(titulo_str: str) -> bool:
    # Clean non-digits
    te = re.sub(r'\D', '', titulo_str)
    
    # Must be exactly 12 digits and cannot be a repeated sequence (e.g., 111111111111)
    if len(te) != 12 or te == te[0] * 12:
        return False

    # Digits 9 and 10 represent the UF (State Code). Must be between 01 (SP) and 28 (Exterior)
    uf = int(te[8:10])
    if uf < 1 or uf > 28:
        return False

    try:
        # Calculate First Check Digit (DV1) - Weights 2 through 9
        sum1 = sum(int(te[i]) * (i + 2) for i in range(8))
        rem1 = sum1 % 11
        
        # Rule exception for SP (01) and MG (02)
        if rem1 == 10:
            dv1 = 0
        elif rem1 == 0:
            dv1 = 1 if uf in (1, 2) else 0
        else:
            dv1 = rem1
            
        # Calculate Second Check Digit (DV2) - Weights 7, 8, 9 applied to UF and DV1
        sum2 = (int(te[8]) * 7) + (int(te[9]) * 8) + (dv1 * 9)
        rem2 = sum2 % 11
        
        if rem2 == 10:
            dv2 = 0
        elif rem2 == 0:
            dv2 = 1 if uf in (1, 2) else 0
        else:
            dv2 = rem2
            
        # Compare calculated digits with the last two digits of the input
        return te[-2:] == f"{dv1}{dv2}"
        
    except ValueError:
        return False

# 3. Extract, Validate, and Mask
def extract_valid_titulos(text: str) -> list[str]:
    if not isinstance(text, str):
        return []
    
    matches = re.findall(titulo_candidate_regex, text)
    valid_titulos = []
    
    for candidate in matches:
        raw_digits = re.sub(r'\D', '', candidate)
        if is_valid_titulo(raw_digits):
            # Extract the UF code to map the state (optional context, good for audits)
            uf_code = raw_digits[8:10]
            # Mask all but the last 4 digits
            masked = f"Título de Eleitor (UF: {uf_code}): ********{raw_digits[-4:]}"
            valid_titulos.append(masked)
            
    return valid_titulos

for col in text_cols:
    df[f'TITULO_FOUND_IN_{col}'] = df[col].apply(extract_valid_titulos)

# Filter rows where at least one valid Título was discovered
mask_titulo = df[[f'TITULO_FOUND_IN_{col}' for col in text_cols]].apply(lambda row: any(len(c) > 0 for c in row), axis=1)
df_titulo_found = df[mask_titulo].copy()

# Display results
print(f"Found {len(df_titulo_found)} rows containing Títulos de Eleitor.")
display_cols_titulo = ['SOLICITACAO'] + [f'TITULO_FOUND_IN_{col}' for col in text_cols]
display(df_titulo_found[display_cols_titulo])
```

```python
import re
import pandas as pd

# 1. Regex to capture 11-digit sequences with optional dots, hyphens, or spaces
pis_candidate_regex = r'\b(?:\d[.\-\s]*){11}\b'

# 2. PIS / PASEP / NIS / NIT Modulo 11 Validator
def is_valid_pis(pis_str: str) -> bool:
    pis = re.sub(r'\D', '', pis_str)
    
    # Must be 11 digits and not an obvious placeholder (e.g., 00000000000)
    if len(pis) != 11 or pis == pis[0] * 11:
        return False
    
    # Official CNIS Modulo 11 Weights
    weights = [3, 2, 9, 8, 7, 6, 5, 4, 3, 2]
    
    try:
        total_sum = sum(int(digit) * weight for digit, weight in zip(pis[:10], weights))
        remainder = total_sum % 11
        
        # If remainder is 0 or 1, check digit is 0; otherwise 11 - remainder
        expected_dv = 0 if remainder < 2 else 11 - remainder
        
        return int(pis[10]) == expected_dv
    except ValueError:
        return False

# 3. Extract, Validate, and Mask
def extract_valid_pis(text: str) -> list[str]:
    if not isinstance(text, str):
        return []
    
    matches = re.findall(pis_candidate_regex, text)
    valid_pis_list = []
    
    for candidate in matches:
        raw_digits = re.sub(r'\D', '', candidate)
        if is_valid_pis(raw_digits):
            # Mask the registration number, revealing only the last 3 digits
            masked = f"PIS/NIS: ********{raw_digits[-3:]}"
            valid_pis_list.append(masked)
            
    return valid_pis_list


for col in text_cols:
    df[f'PIS_FOUND_IN_{col}'] = df[col].apply(extract_valid_pis)

# Filter rows where at least one valid PIS was discovered
mask_pis = df[[f'PIS_FOUND_IN_{col}' for col in text_cols]].apply(lambda row: any(len(c) > 0 for c in row), axis=1)
df_pis_found = df[mask_pis].copy()

# Display results
print(f"Found {len(df_pis_found)} rows containing PIS/PASEP/NIS numbers.")
display_cols_pis = ['SOLICITACAO'] + [f'PIS_FOUND_IN_{col}' for col in text_cols]
display(df_pis_found[display_cols_pis])
```

# Name - spacy

```python
import pandas as pd
import spacy
from tqdm import tqdm

# 1. Use the large model and keep the parser (we only disable lemmatizer/attribute_ruler)
nlp = spacy.load("pt_core_news_sm", disable=["attribute_ruler", "lemmatizer"])

df_nlp = df.copy()

# 2. Domain-specific blocklist. Add words here that commonly appear as false positives.
# Keep them lowercase for easy matching.
BLOCK_WORDS = {
    "solicita", "fiscalização", "trânsito", "devido", "ambulância", 
    "informa", "marca", "chat", "frente", "praça", "rua", "avenida", 
    "veículo", "duster", "branca", "aloja", "aliança", "novo", "mundo"
}

def is_valid_name(name_string):
    """Checks if the extracted name is actually valid."""
    # Convert to lowercase to check against blocklist
    tokens = set(name_string.lower().split())
    
    # If any word in the name is in our blocklist, reject it
    if tokens.intersection(BLOCK_WORDS):
        return False
        
    # Reject single letters or extremely short names
    if len(name_string) <= 2:
        return False
        
    return True

def prep_text(text):
    if not isinstance(text, str):
        return ""
    # .title() is dangerous, but if we must recover case from ALL CAPS, we keep it 
    # and rely heavily on the is_valid_name() filter to clean up the mess it makes.
    return text.title() if text.isupper() else text

for col in text_cols:
    print(f"\nStarting NLP extraction for column: {col}")
    
    texts = [prep_text(t) for t in df_nlp[col]]
    
    results = []
    # n_process=-1 uses all cores. 
    for doc in tqdm(nlp.pipe(texts, n_process=-1, batch_size=500), total=len(texts), desc=f"Processing {col}"):
        
        # 3. Extract PER entities AND filter them through our validation function
        valid_names = [
            ent.text for ent in doc.ents 
            if ent.label_ == "PER" and is_valid_name(ent.text)
        ]
        results.append(valid_names)
        
    df_nlp[f'NAMES_FOUND_IN_{col}'] = results

# Filter rows where at least one list is not empty
mask_nlp = df_nlp[[f'NAMES_FOUND_IN_{col}' for col in text_cols]].apply(
    lambda row: any(len(x) > 0 for x in row), axis=1
)
df_names_nlp = df_nlp[mask_nlp]

print(f"\nFound {len(df_names_nlp)} rows containing potential names.")
display_cols_names = ['SOLICITACAO'] + [f'NAMES_FOUND_IN_{col}' for col in text_cols]
display(df_names_nlp[display_cols_names])
```
